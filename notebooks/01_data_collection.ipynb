Import required libraries

```python
import yfinance as yf
import pandas as pd
from pathlib import Path
```

Defining stock

Initially preparing model on one stock later expanding into many
Initially collecting historical data for Reliance Industries (NSE).

```python
STOCK = "RELIANCE.NS"
START_DATE = "2015-01-01"
```

Downloading using yfinance

```python
df = yf.download(STOCK, start=START_DATE, auto_adjust=False)

df.head()
```

Clean the column structure and reset the date index

```python
df.dropna(inplace=True)
```

```python
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df = df.reset_index()

df.head()
```

Save the original data as a raw CSV

```python
raw_path = Path("../data/raw/RELIANCE.csv")

df.to_csv(raw_path,index=False)
print(f"Raw data saved to: {raw_path}")
```
